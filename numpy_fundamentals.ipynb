## What is numpy?

NumPy (short for Numerical Python) is a fundamental Python library for scientific computing. It provides support for large, multi-dimensional arrays and matrices, along with a vast collection of high-level mathematical functions to operate on these arrays efficiently.

## Numpy vs Python Sequences

- Numpy optimized in C (100x faster), supports vectorized operations (no loops). Lists slower (Python-interpreted), requires loops for math operations.
- NumPy homogeneous (fixed dtype), contiguous memory (efficient for large data). Lists heterogeneous (mixed types), dynamic but memory-heavy (stores pointers).
- NumPy built-in math (np.sin(), np.dot()), broadcasting, multi-dimensional support. Lists no native math ops, no broadcasting, only 1D (nested lists are messy).
- NumPy numerical computing (ML/data science), large datasets, matrix operations. Lists general-purpose tasks, mixed data types, frequent resizing (append/pop).

## Indexing and Slicing

```python
a9 = np.arange(10)
a10 = np.arange(12, dtype=float).reshape(3,4)
a11 = np.arange(27).reshape(3,3,3)
```

```python
## 1D Array Indexing
print(a9)
print(a9[2])
```

```python
## 1D Array Reverse Indexing
print(a9[-1])
```

```python
## 2D Array Indexing
print(a10)
print(a10[2,2])
```

```python
## 3D Array Indexing
print(a11)
print(a11[1,0,1])
```

```python
## 1D Array Slicing
print(a9)
print(a9[2:5])
```

```python
## 1D Array Reverse Slicing
print(a9[-1:-4:-1])
```

```python
## 2D Array Slicing Row
print(a10)
print(a10[1,:])
```

```python
## 2D Array Slicing Column
print(a10)
print(a10[:,1])
```

```python
## 2D Array Slicing Multiple Rows, Columns
print(a10)
print(a10[0:2,2:])
print(a10[::2,1::2])
print(a10[1:2,::3])
print(a10[:2,1::])
```

```python
## 3D Array Slicing
print(a11)
# print(a11[1,:,1])
# print(a11[2,1:,1:])
print(a11[::2,0:1,::2])
```
