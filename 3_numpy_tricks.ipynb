## Sort

```python
## np.sort()
import numpy as np

arr1 = np.random.randint(1,100,50)
## asc
np.sort(arr1)

## dec
np.sort(arr1)[::-1]
```

## Append

```python
## 1D
np.append(arr1, 500)
```

```python
## 2D
arr2 = np.random.randint(1,20,12).reshape(4,3)
np.append(arr2, np.int64(np.random.random((arr2.shape[0], 2))*100), axis=1)
```

## Concate

```python
## concatenate
arr3 = np.random.randint(1,20,12).reshape(4,3)
arr4 = np.random.randint(1,20,12).reshape(4,3)
np.concatenate((arr3, arr4), axis=1)
```

## Unique

```python
## unique
arr5 = np.array([1,2,3,4,5,6,6,6,6,7,6,4,5,6])
np.unique(arr5)
```

## Expand Dimension

```python
## expand_dim
arr6 = np.arange(12)
np.expand_dims(arr6, axis=1)
```

## Where

```python
## where => np.where(condition, true_value, false_value)
arr7 = np.random.randint(1, 100, 20)
np.where(arr7>20, 10, arr7)
```

## Arg max, min

```python
## max
arr8 = np.random.randint(1, 100, 20)
np.argmax(arr8)
```

```python
## min
arr8 = np.random.randint(1, 100, 20)
np.argmin(arr8)
```

## Cumulative

```python
## cumsum
arr9 = np.array([1,2,3,4,5,6,6,6,6,7])
np.cumsum(arr9)
```

```python
## cumprod
np.cumprod(arr9)
```

## Percentile

```python
## percentile
np.percentile(arr9, 10)
```

## More built in functions

```python
## histogram
np.histogram(arr9, bins=[0,2,4,6,8,10])
```

```python
## corrcoef
arr10 = np.random.randint(1,100,10)
arr11 = np.random.randint(1,100,10)
np.corrcoef(arr10,arr11)
```

```python
## isin
arr10[np.isin(arr10, arr11)]
```

```python
## flip
np.flip(arr11)
```

```python
## put => it will change the given array permanently => np.put(array, [index, idex], [replace_value, r_v)])
arr12 = np.array([3,45,5,6,7,8,54,3])
np.put(arr12, [1,6], [4,5])
arr12
```

```python
## delete => np.delete(arr, [arr_index])
np.delete(arr12, 5)
```

## Set functions

```python
## union1d
np.union1d(arr10, arr11)
```

```python
## intersect1d
np.intersect1d(arr10, arr11)
```

```python
## setxor1d
np.setxor1d(arr10, arr11)
```

```python
## in1d
arr10[np.in1d(arr10, arr11)]
```

```python
## clip => it will return given range numbers
np.clip(arr10, a_min=10, a_max=50)
```
