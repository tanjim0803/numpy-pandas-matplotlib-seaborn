## Filtering a DataFrame

```python
## find all the final winners
final_match = ipl['MatchNumber'] == 'Final'
winning_team = ipl[final_match]['WinningTeam'] == 'Chennai Super Kings'
winning_team.shape[0]
```

```python
## how many super over finishes have occured
total_super_over = ipl[ipl['SuperOver'] == 'Y'].shape[0]
total_super_over
```

```python
## how many matches has csk won in kolkata
city = ipl[ipl['City'] == 'Kolkata']
csk = city['WinningTeam']
```

```python
## toss winner is match winner in percentage
toss = ipl['TossWinner'] == ipl['WinningTeam']
match_winning_team = ipl[toss].shape[0]
percentage = (match_winning_team/ipl.shape[0]) * 100
percentage
```

```python
## movies with rating higher than 8 and votes>10000
rating_votes = (movies_info['imdb_rating'] >= 8) & (movies_info['imdb_votes'] >= 10000)
movies_info[rating_votes].shape[0]
```

```python
## Action movies with rating higher than 7.5
## genre = movies_info['genres'].str.split('|').apply(lambda x:'Action' in x)
genre = movies_info['genres'].str.contains('Action')
rating = (movies_info['imdb_rating'] > 7.5) & genre
movies_info[rating].shape[0]
```
