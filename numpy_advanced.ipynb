## Numpy Array vs Python Lists

#### Speed

```python
## list
a = [i for i in range(10000000)]
b = [i for i in range(10000000, 20000000)]

c = []

import time

start = time.time()

for i in range(len(a)):
    c.append(a[i] + b[i])

print(time.time() - start)
```

```python
## numpy
import numpy as np

arr1 = np.arange(10000000)
arr2 = np.arange(10000000, 20000000)
start = time.time()
arr3 = arr1 + arr2

print(time.time() - start)
```

### Memory

```python
## list
a2 = [i for i in range(10000000)]

import sys

sys.getsizeof(a2)
```

```python
## numpy
arr4 = np.arange(10000000, dtype=np.int8)

sys.getsizeof(arr4)
```

## Advanced Indexing

```python
## Normal Indexing and slicing

arr5 = np.arange(24).reshape(6,4)
arr5[1,2]
```

```python
## Fancy Indexing
print(arr5[[0,2,3,5]])
print(arr5[:, [0,2,3]])
```

```python
## Boolean Indexing
arr6 = np.random.randint(1,100, 24).reshape(6,4)
arr6
```

```python
## find all numbers greater than 50
arr6[arr6>50]
```

```python
## find all numbers greater than 50 and are even
arr6[(arr6>50) & (arr6%2==0)]
```

```python
## find all number not divisible by 7
arr6[~(arr6%7==0)]
```

## Broadcasting

Broadcasting is a powerful NumPy feature that allows arithmetic operations between arrays of different shapes by automatically expanding smaller arrays to match larger ones without copying data.

```python
## same shape
rand_arr1 = np.int64(np.random.random(12)*100).reshape(4,3)
rand_arr2 = np.int64(np.random.random(12)*100).reshape(4,3)

rand_arr3 = rand_arr1 + rand_arr2
rand_arr3
```

```python
## diff shape
rand_arr4 = np.int64(np.random.random(12)*100).reshape(4,3)
rand_arr5 = np.int64(np.random.random(3)*100).reshape(1,3)

print(rand_arr4+rand_arr5)
```

## Working with mathematical formulas

```python
## sigmoid
def sigmoid(arr):
    return 1/(1+np.exp(-(arr)))

arr7 = np.arange(10)

sigmoid(arr7)
```

## Working with missing values

```python
## Working with missing values -> np.nan
arr8 = np.array([1,2,3,4,np.nan,5])

arr8[~np.isnan(arr8)]
```
