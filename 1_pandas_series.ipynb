## What is pandas?
Pandas is a powerful, open-source data analysis and manipulation library for Python. It's one of the most widely used tools in data science and analytics.

## Pandas Series
A Pandas Series is like a column in a table. It is a 1-D array holding data of any type.

```python
## Importing pandas
import pandas as pd
import numpy as np

```

## Series from lists

```python
# string
country = ["Bangladesh", "Pakistan", "Nepal", "Bhutan"]
pd.Series(country)
```

```python
# integers
numbers = [1,3,4,6,7,8]
pd.Series(numbers)
```

```python
# custom index
marks = [67,57,89,100]
subjects = ['maths','english','science','bangla']

pd.Series(marks,index=subjects)
```

```python
# setting a name
marks = pd.Series(marks,index=subjects,name='Tanjim Marks')
marks
```

```python
## series from dict
marks = {
    'maths':67,
    'english':57,
    'science':89,
    'bangla':100
}

marks_series = pd.Series(marks,name='Tanjim Marks')
marks_series
```

## Series Attributes

```python
# size
marks_series.size
```

```python
# dtype
marks_series.dtype
```

```python
# is_unique
marks_series.is_unique

pd.Series([1,1,2,3,4,5]).is_unique
```

```python
# index
marks_series.index
```

```python
# values
marks_series.values
```
